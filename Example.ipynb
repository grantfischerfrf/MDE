Run inference and save data:
This script can run inference on various types of data: 
VIDEO: Takes .MOV or .MP4 files, splits into frames of desired fps, and runs inference on each frame.
    input structure: A list of files paths
TOWERFRAMES: Takes .tiff files from the Jack-up Tower. Runs inference on each frame. Searches for folders inside input path with a date in format: YYYMMDD*. Images are in format Camera_*_UnixTimestamp.tiff
(ex: CamBob_A_1685988479999)
    input structure: File path to a folder containing subfolders named by date. Each subfolder contains images.
RAW IMAGES: Opens .npz files with dictionaries containing raw image arrays. Runs inference on each array. 
    input structure: A list of file paths to .npz files.
