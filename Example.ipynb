## **Example Script**

The example data is sourced from RGB and NIR cameras mounted on the LARC. The purpose of this script is to compare the performance of MDE models to Stereo depth estimation

**Imports:**

```python
import torch
import os
import glob
import models
import depth_utils
import depth_plots

%matplotlib inline
```

**Select device:**

```python
device = "cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu"
```

**Select MDE model:**

 For the example data, Depth_Anything_V2 (metric) is the preferred model. All models output estimations in metric depth. Depth_Anything_V2 can create outputs in both relative and metric depth. Use flag pred='metric' or pred='relative' to select the output type.

```python
model = models.dep_any(device, pred='metric')
# model = models.glpn(device)
# model = models.intel_zoe(device)
# model, transform = models.dep_pro(device)
```

**Run inference and save data:**

*run_model can be: 'dep_any', 'glpn', 'dpt_zoe', 'dep_pro'*

RUN RAW IMAGES: Opens .npz files with dictionaries containing raw image arrays. Runs inference on each array. Saves output.

*input structure*: A list of file paths to .npz files containing data.

```python
'''RUN RAW IMAGES'''
input_path = [
    './data/ms_output_1/ms2/left_image_rect/ms_output_1_ms2_left_image_rect.npz',
    './data/ms_output_1/ms2/aux_image_rect_color/ms_output_1_ms2_aux_image_rect_color.npz',
    './data/ms_output_1/ms1/left_image_rect/ms_output_1_ms1_left_image_rect.npz',
    './data/ms_output_2/ms2/left_image_rect/ms_output_2_ms2_left_image_rect.npz',
    './data/ms_output_2/ms2/aux_image_rect_color/ms_output_2_ms2_aux_image_rect_color.npz',
    './data/ms_output_2/ms1/left_image_rect/ms_output_2_ms1_left_image_rect.npz',
]
data_name = [f'{os.path.basename(path).split(".")[0]}' for path in input_path]
depth_utils.run_rawImages(model, input_path, device, run_model='dep_any', data_name=data_name)
```

### **Plot data:**

plot_utils.py contains numerous ways to visualize data - currently most functionality is for analysis of the example data

flag = 'RGB' or 'NIR' to select the type of data to plot.

dataset = 'ms_output_#' to select desired dataset

all_data = True or False to select whether to plot the entire dataset or individual frames.

**error metrics all data:** returns error metrics for entire dataset

```python
depth_plots.error_metrics_allData(data_folder='./Depth_Anything_V2/data', output_path='./Depth_Anything_V2/outputs/temp/', dataset='ms_output_1')
```

**error metrics:** returns error metrics for a single frame

```python
depth_plots.error_metrics(data_folder='./Depth_Anything_V2/data', output_path='./Depth_Anything_V2/outputs/temp/', dataset='ms_output_1')
```

**depth comparison:** returns the comparison between MDE and Stereo depth estimation for a single frame. Includes RMSE and RMSPE.

```python
depth_plots.depth_comparison(data_folder='./Depth_Anything_V2/data', output_path='./Depth_Anything_V2/outputs/temp/', flag='RGB', dataset='ms_output_1')
```

**one-to-one:** returns a one-to-one plot comparing MDE and Stereo depth estimation at each pixel in an image. 

```python
depth_plots.one_to_one(data_folder='./Depth_Anything_V2/data', output_path='./Depth_Anything_V2/outputs/temp/', flag='RGB', dataset='ms_output_1', all_data=True)
```

**histogram:** returns a histogram showing the error distribution between MDE and Stereo depth estimation. 

```python
depth_plots.histogram(data_folder='./Depth_Anything_V2/data', output_path='./Depth_Anything_V2/outputs/temp/', flag='RGB', dataset='ms_output_1', all_data=True)
```

**heatmap:** returns a heatmap showing the density of points in the one-to-one plot. 

```python
depth_plots.heatmap(data_folder='./Depth_Anything_V2/data', output_path='./Depth_Anything_V2/outputs/temp/', flag='RGB', dataset='ms_output_1', all_data=True)
```

**fill between plot:** returns a fill-between plot showing error between MDE and Stereo depth estimation. 

```python
depth_plots.fill_between_plot(data_folder='./Depth_Anything_V2/data', output_path='./Depth_Anything_V2/outputs/temp/', flag='RGB', dataset='ms_output_1', all_data=True)
```

**fill between overlay:** returns a fill-between plot showing error between MDE and Stereo depth estimation for RGB and NIR data.

```python
depth_plots.fill_between_overlay(data_folder='./Depth_Anything_V2/data', output_path='./Depth_Anything_V2/outputs/temp/', dataset='ms_output_1', all_data=True)
```

**one-to-one overlay:** returns a one-to-one plot comparing MDE and Stereo depth estimation for RGB and NIR data. 

```python
depth_plots.one_to_one_overlay(data_folder='./Depth_Anything_V2/data', output_path='./Depth_Anything_V2/outputs/temp/', dataset='ms_output_1', all_data=True)
```

**histogram stack:** returns a stacked histogram showing the error distribution between MDE and Stereo depth estimation for RGB and NIR data. 

```python
depth_plots.histogram_stack(data_folder='./Depth_Anything_V2/data', output_path='./Depth_Anything_V2/outputs/temp/', dataset='ms_output_1', all_data=True)
```
